# Plotting of 

```python
df_err = df.groupby(["Method", "Model"]).agg({
    "is_correct": "mean",
}).reset_index()
# pivot to have Method as Columns
df_err = df_err.pivot_table(index=["Model"], columns="Method", values="is_correct", fill_value=np.nan).reset_index()
df_err.head(50)
```

```python
df_t = (
    df.groupby(["Method", "Model"])
    .agg({
        "is_correct": ["sum", "mean"]
    })
    .reset_index()
    .sort_values(by=("is_correct", "sum"), ascending=False)
    .reset_index(drop=True)
)
df_t
```

```python
mean_token_usage = df.groupby(["Model", "Method"]).agg(
    mean_correct_answers=("is_correct", "mean"),
    mean_input_tokens=("input_tokens", "mean"),
    mean_output_tokens=("output_tokens", "mean"),
    mean_total_tokens=("total_tokens", "mean"),
)
mean_token_usage
```
