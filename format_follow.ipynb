## Output Parser

```python
import re
from pydantic import BaseModel, Field
from typing import List, Literal
from langchain.output_parsers import RetryOutputParser, PydanticOutputParser
from langchain.prompts import PromptTemplate
from langchain_ollama import ChatOllama

# --- Pydantic schemas ---

class Step(BaseModel):
    plan: str = Field(
        description="A short, clear reasoning statement describing the immediate action to take and why it is needed — not the full solution."
    )
    tool: Literal["Calculator"] = Field(
        description="The exact name of the tool to be used for this step (must match an available tool name)."
    )
    tool_input: str = Field(
        description="Mathematical expression in SymPy-compatible syntax for evaluation."
    )

class Plan(BaseModel):
    steps: List[Step] = Field(
        description="An ordered list of steps, where each step contains its reasoning plan, chosen tool, and the exact input for that tool."
    )


# --- Custom parser that strips <think> blocks before parsing ---

class NoThinkOutputParser(PydanticOutputParser):
    def parse(self, text: str):
        text = re.sub(r"<think>.*?</think>", "", text, flags=re.DOTALL)
        return super().parse(text)


# --- Instantiate parser and LLM ---

# Note: 'model' variable needs to be defined
model = "qwen3:0.6b"  # or whatever model you're using

plan_parser = NoThinkOutputParser(pydantic_object=Plan)

llm = ChatOllama(model=model, temperature=0, num_predict=1000, seed=42)

retry_parser = RetryOutputParser.from_llm(
    parser=plan_parser,
    llm=llm,
)

# --- Prepare prompt template with format instructions ---

format_instructions = plan_parser.get_format_instructions()

template = """
You are the Planner in the ReWOO framework.
Break the question into steps and output JSON matching the schema.

{format_instructions}

Question: {query}
Response:
"""

prompt = PromptTemplate(
    template=template,
    input_variables=["query"],
    partial_variables={"format_instructions": format_instructions}
)
```

```python
query = "A baker makes 25 muffins, sells 10, then bakes 15 more. How many muffins does he have now?"
query, answer = SVAMP.get(1, "test")

# Format the prompt first
formatted_prompt = prompt.format(query=query)

# Get raw output from LLM
raw_output = llm.invoke(formatted_prompt)
print(raw_output)
try:
    plan = retry_parser.parse_with_prompt(raw_output.content, formatted_prompt)
except Exception as e:
    print(f"Parsing failed: {e}")
    # Fallback to regular parser
    try:
        plan = plan_parser.parse(raw_output)
    except Exception as e2:
        print(f"Regular parsing also failed: {e2}")
        print(f"Raw output was: {raw_output}")
        raise
print(query)
print(answer)
print(plan)
```

```python
import re
from pydantic import BaseModel, Field
from typing import List, Literal
from langchain.output_parsers import RetryOutputParser, PydanticOutputParser
from langchain.prompts import PromptTemplate
from langchain_ollama import ChatOllama

# --- Pydantic schemas ---

class Step(BaseModel):
    plan: str = Field(
        description="A short, clear reasoning statement describing the immediate action to take and why it is needed — not the full solution."
    )
    equation_variable : str = Field(
        description="The variable name for the equation result, e.g., #E1, #E2, etc. This is used to reference the result in subsequent steps."
    )
    tool: Literal["Calculator"] = Field(
        description="The exact name of the tool to be used for this step (must match an available tool name)."
    )
    tool_input: str = Field(
        description="Mathematical expression in SymPy-compatible syntax for evaluation. It can contain plan variables such as #E1, #E2, etc., which will be replaced with their values during execution."
    )

class Plan(BaseModel):
    steps: List[Step] = Field(
        description="An ordered list of steps, where each step contains its reasoning plan, chosen tool, and the exact input for that tool."
    )


# --- Custom parser that strips <think> blocks before parsing ---

class NoThinkOutputParser(PydanticOutputParser):
    def parse(self, text: str):
        text = re.sub(r"<think>.*?</think>", "", text, flags=re.DOTALL)
        return super().parse(text)


# --- Instantiate parser and LLM ---

# Note: 'model' variable needs to be defined
model = "qwen3:0.6b"  # or whatever model you're using

plan_parser = NoThinkOutputParser(pydantic_object=Plan)

llm = ChatOllama(model=model, temperature=0, num_predict=1000, seed=42)

retry_parser = RetryOutputParser.from_llm(
    parser=plan_parser,
    llm=llm,
)

# --- Prepare prompt template with format instructions ---

format_instructions = plan_parser.get_format_instructions()

template = """
You are the Planner in the ReWOO framework and are given the plan to solve the question.
Bring the plan into JSON matching the schema and fix any format issues with the plan such as missing plan definitions, incorrect tool names, or wrong tool inputs. 
Remember, each equation (#E1, #E2, etc.) must follow after a "Plan:" definition as follows:

Plan: Calculate the initial total number of fish by adding the goldfish and catfish. #E1 = Calculator[7 + 12]

If there is a missing plan definition, you must add it for the schema.

{format_instructions}

Plan: {user_plan}
Response:
"""

prompt = PromptTemplate(
    template=template,
    input_variables=["user_plan"],
    partial_variables={"format_instructions": format_instructions}
)
```

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_ollama import ChatOllama

model = ChatOllama(model="qwen2.5:0.5b", temperature=0, num_predict=2000, seed=42)

prompt = """
You are a helpful assistant that solves math word problems by breaking them down into small, logical steps. Your goal is to generate a sequence of calculations using a provided Calculator tool, leading to the final answer without stating it directly.

Tools available:
(1) Calculator[input]
Performs basic arithmetic operations (addition, subtraction, multiplication, division).
Example usage: Calculator[180 / 3] or Calculator[12 * 4]

**Output Rules:**
1.  **Each step MUST consist of exactly two parts:**
    * **Plan:** A concise explanation of *what specific calculation* will be performed in this step and *why* it's needed for the problem.
    * **Tool call**: A Calculator[...] expression to compute.
2.  **Use ONLY ONE Tool call per step.**
3.  **Save results as unique, sequential variables:** Use #E1 for the first result, #E2 for the second, and so on (e.g., #E1, #E2, #E3...). DO NOT reuse variable names.
4.  **Crucially: Only output the "Plan:" and "#E# = Calculator[...]" lines.** Do NOT include any other text, reasoning, or the final answer.
5.  **Think step by step.** Each step should build upon previous results if necessary. The "Plan:" should describe the calculation in the current step, not the overall problem goal.
6.  **Reference an intermediate result**: Use the "#E#" to reference an intermediate result when using a tool like "Calculator[#E1 + 5]"

**Step Format:**

Plan: [what specific calculation or reasoning this step performs and why]. #E1 = Calculator[math expression]

Example Problem and Solution (for demonstration only – do NOT include this in your actual response):
Task: A baker makes 25 muffins. He sells 10. Then he bakes 15 more. How many muffins does he have now?

Plan: Calculate the number of muffins remaining after sales by subtracting sold muffins from the initial amount. #E1 = Calculator[25 - 10]
Plan: Add the newly baked muffins to the remaining muffins. #E2 = Calculator[#E1 + 15]

Begin! 
Describe your plans with rich details. Each Plan should be followed by only one #E.

Task: {task}
Response:
"""
prompt_template = ChatPromptTemplate.from_messages([("user", prompt)])
planner = prompt_template | model
```

```python
correction_prompt = """
You are a helpful assistant that solves math word problems by breaking them down into small, logical steps. Your goal is to generate a sequence of calculations using a provided Calculator tool, leading to the final answer without stating it directly.

Tools available:
(1) Calculator[input]
Performs basic arithmetic operations (addition, subtraction, multiplication, division).
Example usage: Calculator[180 / 3] or Calculator[12 * 4]

**Output Rules:**
1.  **Each step MUST consist of exactly two parts:**
    * **Plan:** A concise explanation of *what specific calculation* will be performed in this step and *why* it's needed for the problem.
    * **Tool call**: A Calculator[...] expression to compute.
2.  **Use ONLY ONE Tool call per step.**
3.  **Save results as unique, sequential variables:** Use #E1 for the first result, #E2 for the second, and so on (e.g., #E1, #E2, #E3...). DO NOT reuse variable names.
4.  **Crucially: Only output the "Plan:" and "#E# = Calculator[...]" lines.** Do NOT include any other text, reasoning, or the final answer.
5.  **Think step by step.** Each step should build upon previous results if necessary. The "Plan:" should describe the calculation in the current step, not the overall problem goal.
6.  **Reference an intermediate result**: Use the "#E#" to reference an intermediate result when using a tool like "Calculator[#E1 + 5]"

**Step Format:**

Plan: [what specific calculation or reasoning this step performs and why]. #E1 = Calculator[math expression]

Example Problem and Solution (for demonstration only – do NOT include this in your actual response):
Task: A baker makes 25 muffins. He sells 10. Then he bakes 15 more. How many muffins does he have now?

Plan: Calculate the number of muffins remaining after sales by subtracting sold muffins from the initial amount. #E1 = Calculator[25 - 10]
Plan: Add the newly baked muffins to the remaining muffins. #E2 = Calculator[#E1 + 15]

The given output is invalid. Please correct the following:

Invalid Plan: {plan}
Correction:
"""

correction_prompt = ChatPromptTemplate.from_messages([("user", correction_prompt)])
corrector = correction_prompt | model
```

```python
import re

def validate_plan(output_text):
    plan = re.sub(r'<think>.*?</think>', '', output_text, flags=re.DOTALL)
    regex_pattern = r"Plan:\s*(.+)\s*(#E\d+)\s*=\s*(\w+)\s*\[([^\]]+)\]"
    matches = re.findall(regex_pattern, plan)
    if not matches:
        return False, None
    return True, matches

def generate_plan(question, max_attempts=3):
    output = planner.invoke({"task": question}).content
    valid, parsed = validate_plan(output)
    if valid:
        return parsed  # success
    print("invalid", output)
    for i in range(max_attempts):
        output = corrector.invoke({"plan": output}).content
        valid, parsed = validate_plan(output)
        if valid:
            return parsed
        print("Invalid:", output)
    raise ValueError("Failed to generate a valid plan after multiple attempts")

# Example usage
question = task
plan = generate_plan(question)
print(plan)
```
